```python
import pandas as pd
```

```python
dados = pd.DataFrame({'Profissão': [1, 2, 3, 1, 2, 2, 2, 3, 3, 2, 1, 3]})
```

```python
type(dados)
```

```python
# Conta a qtd de dados,transforma em percentual multiplicando por 100 e salva em uma series
frequencia = dados["Profissão"].value_counts()  # em número
percentual = dados["Profissão"].value_counts(normalize=True) * 100  # em percentual

# Salvar os dados em um dataframe
dist_freq_qualitativa = pd.DataFrame(
    {"Frequencia": frequencia, "Percentual": percentual}
)
# Renomeia os index
dist_freq_qualitativa.rename(
    index={1: "Estático", 2: "Cientísta de Dados", 3: "Programador Python"},
    inplace=True,
)
dist_freq_qualitativa.rename_axis("Profissão", axis="columns", inplace=True)

dist_freq_qualitativa
```
