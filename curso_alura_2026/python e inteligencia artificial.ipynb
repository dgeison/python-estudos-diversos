```python
import os
from dotenv import load_dotenv
from google import genai
from time import time, sleep 

load_dotenv()

secret_key = os.getenv("GEMINI_API_KEY")

if not secret_key:
    raise ValueError("GEMINI_API_KEY não encontrada! Verifique seu arquivo .env")

client = genai.Client(api_key=secret_key)

MODEL = "gemini-2.5-flash"


def generate_with_retry(prompt: str, max_tentativas: int = 5):
    espera = 2
    for tentativa in range(1, max_tentativas + 1):
        try:
            return client.models.generate_content(model=MODEL, contents=prompt)
        except Exception as e:
            msg = str(e)
            if tentativa < max_tentativas and ("503" in msg or "UNAVAILABLE" in msg):
                print(f"Tentativa {tentativa} falhou. Nova tentativa em {espera}s...")
                time.sleep(espera)
                espera *= 2
                continue
            raise


response = generate_with_retry("O que é a inteligência artificial?")
print(response.text)
```

```python
def resumidor_de_emails_usando_groq(lista_de_emails):
    client = Groq()

    lista_de_resumos_emails= []

    for numero, email in enumerate(lista_de_emails):
        completion = client.chat.completions.create(
            model="openai/gpt-oss-20b", 
            messages=[{
                "role":"user",
                "content": f"Resuma em apenas uma linha passando o intuito do email. Segue o email: {email}"
            }]
        )
        lista_de_resumos_emails.append(f"E-mail {numero + 1}: {completion.choices[0].message.content}")
        # print(f"E-mail {numero + 1}: {completion.choices[0].message.content}")
        sleep(1)
    return lista_de_resumos_emails
        
resumos = resumidor_de_emails_usando_groq(email_bodies)
```
